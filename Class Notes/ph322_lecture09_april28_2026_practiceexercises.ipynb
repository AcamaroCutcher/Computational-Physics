### Practice Exercise - Numerical partial derivatives/electric field

A point charge is placed at the origin. The electric potential of the point charge in two dimensions at location $(x,y)$ is 

$$
V(r) = \frac{kq}{(x^2+y^2+z^2)^{1/2}}.
$$

where $k$ is the electric constant and $q$ is the charge of the point particle.  The x-component of the electric field is;

$$
E_x(x,y) = -\frac{\partial V(x,y)}{\partial x}.
$$

Calculate the $E_x(x,y)$ over $-1 \le x \le 1, -1 \le y \le 1$. Set $k=1, q=1$. 

Use a  $100 \times 100$ grid.

#### Tip: ```numpy.meshgrid()```

It is often the case we want to create a grid or matrix of coordinate values. In this problem we want to create an $N\times N$ grid that spans $-1$ $\le x \le 1$, and $-1$ $\le y \le 1$.

One convenient way to do this is with the ```numpy``` function ```meshgrid()```. To use ```meshgrid()```,  first create two 1D arrays to hold the $x$ and $y$ coordinates. 

```
x = np.linspace(-1,1, 100)
y = np.linspace(-1,1, 100)
```
The then pass these arrays to ```meshgrid()``` to creates two 2D arrays that contain the $x$ and $y$ coordinates at each grid point.

```
X,Y = np.meshgrid(x,y)
```
